## 1.&nbsp; What is a Reranker?

So far, our RAG system uses a single-stage process for finding information: it retrieves the `top_k` most similar chunks from the vector store and sends them directly to the LLM. This is fast and effective, but we can improve its precision.

A reranker introduces a second, more sophisticated stage to this process.

1.  **Stage 1: Initial Retrieval (Fast & Broad)**: The vector store quickly retrieves a larger set of potentially relevant documents (e.g., the top 10 or 20). This initial search is optimised for speed.

2.  **Stage 2: Reranking (Slow & Precise)**: The reranker then takes this larger set of documents and uses a more powerful model (typically a cross-encoder) to carefully re-assess and re-order them based on their actual relevance to the question. It then passes only the best few (e.g., the top 2 or 5) to the LLM.

The key benefit is a cleaner, more relevant context for the LLM. By filtering out the noise from the initial retrieval, the reranker helps reduce hallucinations and leads to more accurate, focused answers.

### 2.2 Create the Reranker Evaluation Function

```python
# Add to existing llama-index.core imports
from llama_index.core.query_engine import RetrieverQueryEngine # <-- Add this line
from llama_index.core.postprocessor import SentenceTransformerRerank # <-- Add this line

# Add the new configs to the import from evaluation.evaluation_config
from evaluation.evaluation_config import (
    # ... existing imports
    CHUNKING_STRATEGY_CONFIGS,
    RERANKER_MODEL_NAME, # <-- Add this line
    RERANKER_CONFIGS, # <-- Add this line
)
```

2.  **Add the Evaluation Function**: Add the following function to the file, placing it after the `evaluate_chunking_strategies` function.

```python
def evaluate_reranker_strategies() -> None:
    """
    Evaluates different reranker settings on top of the best chunking strategy.
    """
    print("\n--- 🚀 Stage 3: Evaluating Reranker Strategies ---")

    llm_to_test: Groq = initialise_llm()

    embed_model_to_test: HuggingFaceEmbedding = get_embedding_model()

    questions, ground_truths = get_evaluation_data()

    ragas_llm: LlamaIndexLLMWrapper
    ragas_embeddings: HuggingFaceEmbeddings
    ragas_llm, ragas_embeddings = load_ragas_models()

    index: VectorStoreIndex = get_or_build_index(
        chunk_size=CHUNK_SIZE,
        chunk_overlap=CHUNK_OVERLAP,
        embed_model=embed_model_to_test
    )

    all_results: list[pd.DataFrame] = []

    for config in RERANKER_CONFIGS:

        retriever_k, reranker_n = config['retriever_k'], config['reranker_n']

        print(f"--- Testing Reranker Config: retrieve_k={retriever_k},"
              f" rerank_n={reranker_n} ---")

        retriever = index.as_retriever(similarity_top_k=retriever_k)

        reranker = SentenceTransformerRerank(
            top_n=reranker_n, model=RERANKER_MODEL_NAME
        )

        query_engine = RetrieverQueryEngine.from_args(
            retriever=retriever,
            node_postprocessors=[reranker],
            llm=llm_to_test
        )

        qa_dataset: Dataset = generate_qa_dataset(
            query_engine,
            questions,
            ground_truths
        )

        print("--- Running Ragas evaluation for reranker... ---")

        # --- If you don't have a Rate per Minute limit on your API ---
        # results_df: pd.DataFrame = evaluate_without_rate_limit(
        #     qa_dataset,
        #     ragas_llm,
        #     ragas_embeddings,
        # )

        # --- If you do have a Rate per Minute API limit ---
        results_df: pd.DataFrame = evaluate_with_rate_limit(
            qa_dataset,
            ragas_llm,
            ragas_embeddings,
        )

        results_df['chunk_size'] = CHUNK_SIZE
        results_df['chunk_overlap'] = CHUNK_OVERLAP
        results_df['retriever_k'] = retriever_k
        results_df['reranker_n'] = reranker_n

        all_results.append(results_df)

    final_df: pd.DataFrame = pd.concat(all_results, ignore_index=True)

    save_results(final_df, "reranker_evaluation")

    print("--- ✅ Reranker Strategy Evaluation Complete ---")
```

> **Why the `RetrieverQueryEngine`?**
>
> You'll notice we've switched from the simple `index.as_query_engine()` to the more explicit `RetrieverQueryEngine.from_args()`. This is necessary because a reranker is a type of `node_postprocessor` - a component that runs after the initial retrieval but before the final answer generation. The `RetrieverQueryEngine` gives us the fine-grained control to construct this multi-stage pipeline by explicitly defining the `retriever` and the list of `node_postprocessors`.

### 5.2 Upgrade the Chat Engine

we now want to update our `BaseChatEngine`

Replacing the generic `BaseChatEngine` with the `CondensePlusContextChatEngine` and a `SentenceTransformerRerank` postprocessor is a professional upgrade that shifts your system from a "simple chatbot" to an "advanced RAG pipeline".

The core reason for this change is that a simple `BaseChatEngine` often struggles with conversation history and retrieval accuracy as a chat progresses.



1. **update `get_chat_engine`:** in `src/engine.py` and replace the function `get_chat_engine` with the new code below

```python
def get_chat_engine(
        llm: Groq,
        embed_model: HuggingFaceEmbedding
) -> CondensePlusContextChatEngine:
    """Initialises and returns the main conversational RAG chat engine."""
    vector_index: VectorStoreIndex = get_vector_store(embed_model)

    # 1. Create the Retriever
    retriever = vector_index.as_retriever(similarity_top_k=SIMILARITY_TOP_K)

    # 2. Setup the Reranker
    reranker: SentenceTransformerRerank = SentenceTransformerRerank(
        top_n=RERANKER_TOP_N,
        model=RERANKER_MODEL_NAME
    )

    # 3. Setup Memory
    memory: ChatMemoryBuffer = ChatMemoryBuffer.from_defaults(
        token_limit=CHAT_MEMORY_TOKEN_LIMIT
    )

    # 4. Assemble the advanced Chat Engine
    chat_engine = CondensePlusContextChatEngine(
        retriever=retriever,
        llm=llm,
        memory=memory,
        system_prompt=LLM_SYSTEM_PROMPT,
        node_postprocessors=[reranker]
    )

    return chat_engine
```

**Why this upgrade matters?**

**Standalone Queries**: By condensing history into a single question, the retriever is much more likely to find relevant documents than if it queried with the whole chat history.

**Reranking Precision**: Retrieval using embeddings (like bge-large) is good at finding "similar" chunks, but the SentenceTransformerRerank uses a more expensive Cross-Encoder to verify which of those chunks is truly relevant to the query, moving the best info to the top.

**Memory Management**: The `CondensePlusContextChatEngine` maintains state, allowing it to answer meta-questions like "What was the previous topic we discussed?" effectively.

2. **update main_chat_loop:** in `src/engine.py` replace `BaseChatEngine` inside function `main_chat_loop` with our updated `CondensePlusContextChatEngine`

```python
def main_chat_loop() -> None:
    """Main application loop to run the RAG chatbot."""
    print("--- Initialising models... ---")
    llm: Groq = initialise_llm()
    embed_model: HuggingFaceEmbedding = get_embedding_model()

    chat_engine: CondensePlusContextChatEngine = get_chat_engine(
        llm=llm,
        embed_model=embed_model
    )
    print("--- RAG Chatbot Initialised. ---")
    chat_engine.chat_repl()
```

# Reranker Strategy Evaluation — Treasoria RAG

## Objective

The objective of this experiment was to evaluate whether a cross-encoder reranker could improve the retrieval quality of the Treasoria RAG system.

The reranker was tested on top of the preliminary optimal chunking configuration:

- `CHUNK_SIZE = 1024`
- `CHUNK_OVERLAP = 200`
- Reranker model: `cross-encoder/ms-marco-MiniLM-L-6-v2`
- Generator LLM: `openai/gpt-oss-20b`
- Judge LLM: `openai/gpt-oss-120b`
- Metrics: Faithfulness, Context Precision and Context Recall
- Active evaluation questions: 1

The generation limit was increased from 500 to 1,000 tokens to reduce answer truncation.

## Two-Stage Retrieval Process

The reranker evaluation used a two-stage retrieval pipeline:

1. The embedding retriever collected a broad set of candidate chunks.
2. The cross-encoder reranker compared the question with each candidate and retained only the most relevant chunks for answer generation.

Three configurations were evaluated:

| Retriever k | Reranker n | Description |
|---:|---:|---|
| 10 | 2 | Retrieve 10 candidates and retain the best 2 |
| 10 | 3 | Retrieve 10 candidates and retain the best 3 |
| 20 | 3 | Retrieve 20 candidates and retain the best 3 |

## Results

| Chunk size | Overlap | Retriever k | Reranker n | Faithfulness | Context Precision | Context Recall | Response length |
|---:|---:|---:|---:|---:|---:|---:|---:|
| 1024 | 200 | 10 | 2 | NaN | **1.0000** | **1.00** | 2,457 characters |
| 1024 | 200 | 10 | 3 | NaN | 0.8333 | **1.00** | 3,328 characters |
| 1024 | 200 | 20 | 3 | NaN | **1.0000** | **1.00** | 3,950 characters |

## Analysis

All three configurations achieved a perfect Context Recall score. Therefore, every configuration retained the information required by the reference answer.

The `10 → 3` configuration achieved a lower Context Precision score of `0.8333`. This indicates that the third retained chunk introduced partially irrelevant information.

The `10 → 2` and `20 → 3` configurations both achieved perfect Context Precision and Context Recall. However, `10 → 2` used fewer candidate chunks, passed less context to the generator and produced a shorter, more focused answer.

Manual inspection showed that the `10 → 2` response still covered:

- the core accounting identity;
- all seven control totals;
- internal-transfer elimination;
- credit-card repayment treatment;
- the January 2022 payroll debit;
- employer-contribution treatment.

The `20 → 3` response added more technical and secondary details, including synthetic invoices, table relationships and schema information. These details may exist in the documentation but were not necessary to answer the selected question.

## Comparison with the Best Chunking-Only Strategy

The best chunking-only configuration, `1024/200` with `retriever_k=2`, achieved:

- Faithfulness: `0.9231`
- Context Precision: `1.00`
- Context Recall: `1.00`

The best reranker configuration, `10 → 2`, achieved:

- Faithfulness: unavailable
- Context Precision: `1.00`
- Context Recall: `1.00`

Therefore, the reranker did not produce a measurable improvement in Context Precision or Context Recall for this question because the chunking-only system had already achieved perfect retrieval scores.

The reranker response appeared more complete, but this improvement cannot be attributed exclusively to reranking because the generation limit was also increased from 500 to 1,000 tokens before the reranker experiment.

The current question was already precise and closely matched the vocabulary of the Treasoria documents. The reranker may provide more value for vague, ambiguous or differently worded questions.

## Errors and Technical Limitations

The original configurations used `reranker_n=5`. With chunks of 1,024 tokens, five final chunks, the generated response, the reference answer and the internal Ragas prompt exceeded the Groq judge limit of 8,000 tokens.

One failed request contained 8,036 tokens. The configurations were therefore adjusted to retain no more than three chunks.

Further failures occurred during Faithfulness evaluation:

- some requests exceeded the 8,000-token limit;
- the judge sometimes returned an empty JSON object;
- one response contained an incomplete `NLIStatementOutput`;
- Pydantic could not parse the missing `reason` and `verdict` fields.

The evaluation pipeline originally used `raise_exceptions=True`, which caused the complete experiment to stop when one metric failed. It was changed to `raise_exceptions=False`, allowing failed metrics to be recorded as `NaN` while the remaining metrics and configurations continued.

The reranker model was downloaded successfully and executed locally. A temporary Hugging Face connection interruption was automatically retried and did not stop the experiment.

## Selected Reranker Configuration

The preliminary optimal reranker strategy is:

- `SIMILARITY_TOP_K = 10`
- `RERANKER_TOP_N = 2`
- `RERANKER_MODEL_NAME = "cross-encoder/ms-marco-MiniLM-L-6-v2"`

This configuration was selected because it:

- achieved perfect Context Precision;

Testing `retriever_k=50` was not performed during this stage because the current experiment already showed that increasing retrieval from 10 to 20 did not improve the measured retrieval scores.

A larger candidate set would increase local reranking time without demonstrated benefit for the selected question.

Other reranker models were also not tested because `ms-marco-MiniLM-L-6-v2` already achieved perfect Context Precision and Context Recall with the selected configuration.

These experiments may be revisited later using a larger and more challenging evaluation dataset.

## Limitations

- Only one evaluation question was used.
- The question was already precise and easy to retrieve.
- Faithfulness could not be calculated reliably.
- Answer Correctness was not measured.
- The generation limit changed between the chunking and reranker experiments.
- LLM generation and LLM-based evaluation can vary between executions.
- The results do not yet demonstrate a significant improvement over the chunking-only system.

## Conclusion

The reranker pipeline was successfully implemented and evaluated.

The `10 → 2` configuration provided the best balance between retrieval quality, response focus, latency and computational cost. However, the current experiment does not prove that reranking significantly improves Treasoria because the non-reranked system already achieved perfect Context Precision and Context Recall.

The reranker configuration is therefore considered a preliminary optimum. Its final value should be validated with several Treasoria questions, particularly vague questions and questions that use vocabulary different from the source documents.
